```python
''' Create a pandas.Series from a list of integers [10, 20, 30, 40, 50] and set custom index labels: ['a', 'b', 'c', 'd', 'e'].
sample output:
a    10
b    20
c    30
d    40
e    50
dtype: int64
Element at index 'c': 30                                                        
First 3 elements:                                                               
a    10                                                                        
b    20                                                                         
c    30                                                                         
dtype: int64
Display element at index C

Display first 3 elements
'''
import pandas as p
a=p.Series([10,20,30,40,50])
a.index=['a','b','c','d','e']
print(a)
print(f"Element at index 'c': {a['c']}")
print("First 3 elements:")
a=p.Series([10,20,30])
a.index=['a','b','c']
print(a)
```

```python
'''Given the Series s = [5, 10, 15, 20], perform the following operations:

Multiply all elements by 2.

Add 5 to each element
filter the elements and print only those greater than 10
print avg Mean ,max and sum of given series
sample output:
Multiplied by 2:
 0    10
1    20
2    30
3    40
dtype: int64
Added 5:
 0    10
1    15
2    20
3    25
dtype: int64

Values > 10:
 2    15
3    20
dtype: int64
Mean: 12.5
Max: 20
Sum: 50
'''
import pandas as p
s=p.Series([5,10,15,20])
print("Multiplied by 2:")
print(2*s)
print("Added 5:")
print(5+s)
print("Values > 10:")
for i in range(0,5):
    if s[i]>10:
        print(s[i:4])
        break
a=(sum(s))/len(s)
print("Mean:",a)
print("Max:",max(s))
print("Sum:",sum(s))
```

```python
'''Given a Series s = [1, 2, 3, 4, 5] apply a custom function (lambda) to square each number in the Series.
sort the values in both:
Ascending order.
Descending order.
sample output:
Squared Series:
 0     1
 1     4
 2     9
 3    16
 4    25
dtype: int64
Ascending:
  0     1
 1     4
 2     9
 3    16
 4    25
dtype: int64
Descending:
 4    25
 3    16
 2     9
 1     4
 0     1
dtype: int64
    '''
import pandas as p
a=p.Series([1,2,3,4,5])
print("Squared Series:")
print(a*a)
print("Ascending:")
print(a*a)
s=a*a
print("Descending:")
print(s[::-1])
```
